```python
import pandas as pd

data = {
    "Nama": ["Andi", "Budi", "Citra", "Dina", "Eko", "Fajar", "Gita", "Hani"],
    "Produk": ["Laptop", "Mouse", "Keyboard", "Laptop", "Mouse", "Headset", "Keyboard", "Laptop"],
    "Kategori": ["Elektronik", "Aksesoris", "Aksesoris", "Elektronik", "Aksesoris", "Aksesoris", "Aksesoris", "Elektronik"],
    "Harga": [7000000, 150000, 300000, 6500000, 175000, 450000, 275000, 7200000],
    "Jumlah": [1, 2, 1, 1, 3, 2, 2, 1],
    "Kota": ["Kuningan", "Cirebon", "Kuningan", "Majalengka", "Cirebon", "Kuningan", "Majalengka", "Cirebon"]
}

df = pd.DataFrame(data)
df['Total_Harga'] = df['Harga'] * df['Jumlah']
df
```

```python
#mengelompokan berdasarkan kategori produk
pendapatan_kategori = df.groupby('Kategori')['Total_Harga'].sum()
pendapatan_kategori
```

```python
#menggunakan fungsi multi-Aggregasi (agg())
#jika ingin melihat sekaligus jumlah total transaksi, rata rata harga
#dan total barang terjual. gunakan fungsi .agg():

rekap_kota = df.groupby('Kota').agg({

    'Jumlah' : 'sum',
    'Harga' : 'mean',
    'Total_Harga' : 'sum'
})

print(rekap_kota)
```

```python
#1. hitung produk terlaris
# hitung total jumlah barnag  terjual (jumlah) untuk setiap produk  (Laptop, Mouse, Keyboard, headset)
#Produk apa apa yang memiliki total penjualan unit paling banyak?

produk_terlaris = df.groupby('Produk')['Jumlah'].sum().sort_values(ascending = False)

print(produk_terlaris) # produk yang memiliki total penjualan terbanyak yaitu mouse dengan penjualan 5

print("-"*40)

#2. rata-rata harga perkategori
#hitung rata-rata harga produk untuk masing2 kategori

rata_per_kategori = df.groupby('Kategori')['Harga'].mean().sort_values(ascending = False)
print(rata_per_kategori)

print("-"*40)

#3. kontribusi pendapatan berdasarkan kota
#Gunakan fungsi .agg() untuk mencari total pendapatan (Total_Harga) serta jumlah total transaksi di setiap Kota (Kuningan, Cirebon, Majalengka).
#kota mana yang memberikan total pendapatan terbesar?

pendapatan_kota = df.groupby('Kota').agg({
    'Total_Harga' : 'sum',
    'Jumlah' : 'sum'
}).sort_values(by = 'Total_Harga', ascending = False)

print("--- Rekap Pendapatan Berdasarkan Kota ---")
print(pendapatan_kota) #kota Kuningan yang memberikan total pendapatan terbesar
```
