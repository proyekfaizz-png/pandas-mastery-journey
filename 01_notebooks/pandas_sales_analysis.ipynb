```python
import pandas as pd

# Base Dataset Penjualan Retail
data = {
    "Nama": ["Andi", "Budi", "Citra", "Dina", "Eko", "Fajar", "Gita", "Hani"],
    "Produk": ["Laptop", "Mouse", "Keyboard", "Laptop", "Mouse", "Headset", "Keyboard", "Laptop"],
    "Kategori": ["Elektronik", "Aksesoris", "Aksesoris", "Elektronik", "Aksesoris", "Aksesoris", "Aksesoris", "Elektronik"],
    "Harga": [7000000, 150000, 300000, 6500000, 175000, 450000, 275000, 7200000],
    "Jumlah": [1, 2, 1, 1, 3, 2, 2, 1],
    "Kota": ["Kuningan", "Cirebon", "Kuningan", "Majalengka", "Cirebon", "Kuningan", "Majalengka", "Cirebon"]
}

df = pd.DataFrame(data)

df['Total'] = df['Jumlah'] * df['Harga']

print("========== TOTAL PENDAPATAN ==========")
print(df[['Produk', 'Harga', 'Jumlah', 'Total']])

print("-"*40)
#mencari total pendapatan berdasarkan kota
total_kota = df.groupby('Kota')['Total'].sum().sort_values(ascending=False)

print("Total Pendapatan Kota: ")
print(total_kota)

print("-"*40)
#rata rata harga kategori
rata_kategori = df.groupby('Kategori')['Harga'].mean().sort_values(ascending = False)
print(rata_kategori)

print("-"*40)
print("Total Pendapatan Keseluruhan:", df['Total'].sum())
```

```python
#filtering berdasarkan produk
df_laptop = df[df['Produk'] == "Laptop"]

df_laptop
```

```python

#filter berdasarkan harga > 5 juta
exvensive_items = df[df['Harga'] > 5000000]

exvensive_items
```

```python
#melihat total pendapatan atau jumlah barang yang terjual berdasarkan masing kota

sales_per_kota = df.groupby('Kota')['Jumlah'].sum()
print(sales_per_kota)
```
